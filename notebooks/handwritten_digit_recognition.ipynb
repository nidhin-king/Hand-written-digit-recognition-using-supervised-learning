# Handwritten Digit Recognition using Supervised Learning

A Convolutional Neural Network (CNN) trained on the MNIST dataset to classify
handwritten digits (0-9).

This notebook mirrors the modular `src/` package and walks through the full
pipeline: data loading, preprocessing, model building, training, evaluation
and prediction.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

from src.utils import set_seed, IMG_SHAPE, EPOCHS, BATCH_SIZE, VALIDATION_SPLIT, IMAGES_DIR
from src.preprocess import load_data, preprocess_data, display_dataset_info, show_sample_images
from src.train import build_model, train_model, plot_training_history
from src.evaluate import evaluate_model, print_classification_report, plot_confusion_matrix
from src.evaluate import show_sample_predictions, show_correct_incorrect, load_model

set_seed()
```

## 1. Load the dataset

```python
(x_train, y_train), (x_test, y_test) = load_data()
print('Raw shapes:', x_train.shape, y_train.shape, x_test.shape, y_test.shape)
```

## 2. Preprocessing

```python
x_train, y_train, x_test, y_test = preprocess_data(x_train, y_train, x_test, y_test)
display_dataset_info(x_train, y_train, x_test, y_test)
```

```python
fig = show_sample_images(x_train, y_train, num_samples=25, save_path=IMAGES_DIR / 'sample_images.png')
plt.show()
```

## 3. Split into train / validation

```python
x_train, x_val, y_train, y_val = train_test_split(
    x_train, y_train, test_size=VALIDATION_SPLIT, random_state=42
)
print('train:', x_train.shape, 'val:', x_val.shape)
```

## 4. Build the CNN model

```python
model = build_model(IMG_SHAPE)
model.summary()
```

## 5. Train the model

```python
history = train_model(x_train, y_train, x_val, y_val, epochs=EPOCHS, batch_size=BATCH_SIZE)
```

```python
fig = plot_training_history(history, save_path=IMAGES_DIR)
plt.show()
```

## 6. Evaluate on the test set

```python
model = load_model()
metrics = evaluate_model(model, x_test, y_test)
```

```python
y_true = np.argmax(y_test, axis=1)
y_pred = np.argmax(model.predict(x_test, verbose=1), axis=1)
print_classification_report(y_true, y_pred)
```

```python
fig = plot_confusion_matrix(y_true, y_pred)
plt.show()
```

## 7. Sample and correct/incorrect predictions

```python
fig = show_sample_predictions(model, x_test, y_test, num_samples=10)
plt.show()
```

```python
fig = show_correct_incorrect(model, x_test, y_test)
plt.show()
```

Replace `custom_image_path` with the path to your own handwritten digit image.
